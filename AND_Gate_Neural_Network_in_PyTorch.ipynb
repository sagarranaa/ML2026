```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# step 2 create dataset
X=torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

y=torch.tensor([
     [0.],
    [0.],
    [0.],
    [1.]
])
```

```python
# Step 3 create neural network
# This is Huge
model = nn.Sequential(
    nn.Linear(2, 1),
    nn.Sigmoid()
)
```

```python
# step 4 loss function BCELoss(binary cross entropy loss)
criterion = nn.BCELoss()
```

```python
# step 5 optimizer
optimizer = optim.SGD(
    model.parameters(),
    lr=0.1
)
```

```python
# step 6 Traning Loop
for epoch in range(1000):

    # Forward pass
    predictions = model(X)

    # Loss
    loss = criterion(predictions, y)

    # Clear gradients
    optimizer.zero_grad()

    # Backpropagation
    loss.backward()

    # Update weights
    optimizer.step()

    # Print loss
    if epoch % 100 == 0:
        print(
            f"Epoch {epoch}, Loss: {loss.item()}"
        )
```

```python
# Step 7 test model
with torch.no_grad():
  prediction=model(X)
  print(prediction)
```
